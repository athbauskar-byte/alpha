```python
#Accept Student Name, Roll Number and Marks of the 3 subjects from the user. Calculate the percentage of the marks and display it.
#Display the Subject with Highest and lowest marks.
Student_Name=input('Student Name:')
Roll_No=int(input('Enter roll no.:'))
Math_Marks,Physics_Marks,Chemistry_Marks=map(int,input('Enter Marks').split())
Percentage=(Math_Marks+Physics_Marks+Chemistry_Marks)/3
print('Your overall percentage is',Percentage,'%')
Marks={
    'Math':Math_Marks,
    'Physics':Physics_Marks,
    'Chemistry':Chemistry_Marks
}
for x,y in Marks.items():
    if y==max(Marks.values()):
        print('You scored highest marks in',x,'i.e.',y)

for x,y in Marks.items():
    if y==min(Marks.values()):
        print('You scored lowest marks in',x,'i.e.',y)
```

```python
#Write a python program to print prime numbers between 10 to 99
for i in range(10,100):
    for j in range(2,i):
        if i%j==0:
            break
    else:
        print(i)
```

```python
#Write a program to reverse a given number and print
num=int(input('Enter number'))
reversed_num=0
while num>0:                            #while loop
    a=num%10
    reversed_num=(reversed_num*10)+a
    num//=10
print(reversed_num)
```

```python
#Write a program to find if the given number is palindrome or not
num=int(input('Enter number'))
reversed_num=0
while num>0:
    a=num%10
    reversed_num=(reversed_num*10)+a
    num//=10
if reversed_num==num:
    print('This is a palindrome number')
else:
    print('this is not a palindrome number')
```

```python
#Display prime factors of number. Accept number from user.
number=int(input('Enter Number'))
if number>2:
    factor=2

while factor*factor<=number:
    if number%factor==0:
        print(factor)
        number//=factor
    else:
        factor+=1

else:
    print(number)
```
