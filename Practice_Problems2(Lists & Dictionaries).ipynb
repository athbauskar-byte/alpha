```python
#Accept 20 values from user and save it in list. Perform following operations on it
#a) count similar elements of list and print their index value
list5=list(map(int,input().split()))
list6=[]
for i in list5:
    if i in list6:
        continue
    else:
        list6.append(i)
        
for j in list6:
    if list5.count(j)>1:
        index_list=[i for i,x in enumerate(list5) if x==j]
        print(j,'is repeated',list5.count(j),'times and their indices are:',index_list)

#b) count even and odd values of list
even_list=[]
odd_list=[]
for i in list5:
    if i%2==0:
        even_list.append(i)
    else:
        odd_list.append(i)

print('There are',len(even_list),'even values in the list')
print('There are',len(odd_list),'odd values in the list')

#c) count positive and negative values of list.
negative_count=0
positive_count=0
for i in list5:
    if i<0:
        negative_count+=1
    else:
        positive_count+=1

print('There are',negative_count,'negative values in the list')
print('There are',positive_count,'positive values in the list')
```

```python
#An acronym is a word formed by taking the first letters of the words in a phrase and making a word from them.
#For example, RAM is an acronym for “random access memory.”
#Write a program that allows the user to type in a phrase and then outputs the acronym for that phrase.
#Note: the acronym should be all uppercase, even if the words in the phrase are not capitalized.
full_form=list(input().split())
Acronym=[]
for word in range(len(full_form)):
    a=full_form[word][0].upper()
    Acronym.append(a)

b=Acronym[0]
for j in range(1,len(Acronym)):
    b+=Acronym[j]
print('Acronym for the above full form:',b)
```
